```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
# Step 1: Upload image
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import cv2
from matplotlib import pyplot as plt

# Step 3: Get uploaded filename
image_path = next(iter(uploaded))

# Step 4: Read image
img = cv2.imread(image_path)

# Step 5: Check image
if img is None:
    print("Error: Image not found")
else:
    # Convert BGR → RGB for display
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Get image center
    h, w = img.shape[:2]
    center = (w // 2, h // 2)

    # Step 6: Rotation matrices
    clockwise_matrix = cv2.getRotationMatrix2D(center, -45, 1.0)   # clockwise
    anticlockwise_matrix = cv2.getRotationMatrix2D(center, 45, 1.0) # counter-clockwise

    # Step 7: Rotate images
    clockwise_img = cv2.warpAffine(img_rgb, clockwise_matrix, (w, h))
    anticlockwise_img = cv2.warpAffine(img_rgb, anticlockwise_matrix, (w, h))

    # Step 8: Display results
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.title("Original Image")
    plt.imshow(img_rgb)
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.title("Clockwise Rotation (45°)")
    plt.imshow(clockwise_img)
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.title("Counter-Clockwise Rotation (45°)")
    plt.imshow(anticlockwise_img)
    plt.axis("off")

    plt.show()
```

```
# Step 1: Upload image
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Step 3: Read image in grayscale
path = next(iter(uploaded))
img = cv2.imread(path, 0)

if img is None:
    print("Error loading image")
else:
    # Step 4: Apply Gaussian blur
    blurred = cv2.GaussianBlur(img, (5, 5), 0)

    # Step 5: Create unsharp mask
    mask = cv2.subtract(img, blurred)

    # Step 6: Sharpen image (original + mask)
    sharpened = cv2.add(img, mask)

    # Step 7: Display results
    plt.figure(figsize=(12,4))

    plt.subplot(1,4,1)
    plt.imshow(img, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1,4,2)
    plt.imshow(blurred, cmap="gray")
    plt.title("Blurred")
    plt.axis("off")

    plt.subplot(1,4,3)
    plt.imshow(mask, cmap="gray")
    plt.title("Unsharp Mask")
    plt.axis("off")

    plt.subplot(1,4,4)
    plt.imshow(sharpened, cmap="gray")
    plt.title("Sharpened Image")
    plt.axis("off")

    plt.show()
```

```
# Step 1: Upload image
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Step 3: Read image in grayscale
path = next(iter(uploaded))
img = cv2.imread(path, 0)

if img is None:
    print("Error loading image")
else:
    # Step 4: Apply Gaussian blur
    blurred = cv2.GaussianBlur(img, (5, 5), 0)

    # Step 5: Set boost factor (A ≥ 1)
    A = 1.5

    # High-boost filtering formula
    high_boost = cv2.addWeighted(img, A, blurred, -1, 0)

    # Step 6: Display results
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.imshow(blurred, cmap="gray")
    plt.title("Blurred Image")
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.imshow(high_boost, cmap="gray")
    plt.title("High-Boost Sharpened")
    plt.axis("off")

    plt.show()
```

```
# Step 1: Upload image
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Step 3: Read image in grayscale
path = next(iter(uploaded))
img = cv2.imread(path, 0)

if img is None:
    print("Error loading image")
else:
    # Step 4: Sobel gradients
    sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

    # Step 5: Gradient magnitude
    gradient = cv2.magnitude(sobelx, sobely)
    gradient = cv2.convertScaleAbs(gradient)

    # Step 6: Sharpened image
    sharpened = cv2.add(img, gradient)

    # Step 7: Display results
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.imshow(gradient, cmap="gray")
    plt.title("Gradient Mask")
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.imshow(sharpened, cmap="gray")
    plt.title("Sharpened Image")
    plt.axis("off")

    plt.show()
```

```
from google.colab import files
uploaded = files.upload()

import cv2
import numpy as np
from matplotlib import pyplot as plt

paths = list(uploaded.keys())
print("Number of uploaded files:", len(paths))

if len(paths) < 2:
    print("❌ Upload BOTH images together:")
    print("1) Main image containing watch")
    print("2) watch_template.png")
else:
    img = cv2.imread(paths[0])
    template = cv2.imread(paths[1])

    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray_temp = cv2.cvtColor(template, cv2.COLOR_BGR2GRAY)

    result = cv2.matchTemplate(gray_img, gray_temp, cv2.TM_CCOEFF_NORMED)
    _, max_val, _, max_loc = cv2.minMaxLoc(result)

    h, w = gray_temp.shape
    top_left = max_loc
    bottom_right = (top_left[0] + w, top_left[1] + h)

    cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 3)

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(f"Detected Watch (Confidence: {max_val:.2f})")
    plt.axis("off")
```
